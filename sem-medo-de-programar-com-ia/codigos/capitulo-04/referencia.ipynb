## 1. Definições


```python
"""Solução de referência: valida o título antes de cadastrar."""


def cadastrar_tarefa(tarefas, titulo):
    if not titulo.strip():
        return "Informe um título com pelo menos um caractere que não seja espaço."
    tarefas.append({"titulo": titulo, "concluida": False})
    return "Tarefa cadastrada."


def concluir_tarefa(tarefas, indice):
    if indice < 0 or indice >= len(tarefas):
        return "Tarefa não encontrada."
    tarefas[indice]["concluida"] = True
    return "Tarefa concluída."


def listar_tarefas(tarefas):
    if not tarefas:
        return "Nenhuma tarefa cadastrada."
    linhas = []
    for indice, tarefa in enumerate(tarefas):
        marca = "x" if tarefa["concluida"] else " "
        linhas.append(f"{indice}: [{marca}] {tarefa['titulo']}")
    return "\n".join(linhas)


```

## 2. Demonstração

Na versão inicial, são cadastradas três tarefas. Após a correção, apenas Comprar pão deve ser cadastrada, e sua conclusão deve continuar funcionando.


```python
tarefas = []
print(cadastrar_tarefa(tarefas, "Comprar pão"))
print(cadastrar_tarefa(tarefas, ""))
print(cadastrar_tarefa(tarefas, "   "))
print(listar_tarefas(tarefas))
print(concluir_tarefa(tarefas, 0))
print(listar_tarefas(tarefas))
```

## 3. Casos adicionais

Devem ser cadastrados apenas Comprar pão e o título Ler livro com seus espaços preservados. Entradas vazias ou só de espaçamento não devem modificar a lista. Quantidades esperadas após cada caso: 1, 1, 1, 1, 2.


```python
tarefas = []
for titulo in ("Comprar pão", "", "   ", "\t\n", "  Ler livro  "):
    print("Entrada:", repr(titulo))
    print(cadastrar_tarefa(tarefas, titulo))
    print("Quantidade:", len(tarefas))
    print("Títulos:", [repr(tarefa["titulo"]) for tarefa in tarefas])
    print()
```

## 4. Conferência automática

AssertionError indica uma falha. Estudaremos essa técnica no capítulo sobre testes.

```python
"""Execute com Python para conferir os critérios da solução de referência."""

from copy import deepcopy



def verificar():
    tarefas = []
    assert listar_tarefas(tarefas) == "Nenhuma tarefa cadastrada."
    assert cadastrar_tarefa(tarefas, "Comprar pão") == "Tarefa cadastrada."
    assert tarefas == [{"titulo": "Comprar pão", "concluida": False}]
    for titulo in ("", "   ", "\t\n"):
        antes = deepcopy(tarefas)
        assert cadastrar_tarefa(tarefas, titulo).startswith("Informe um título")
        assert tarefas == antes, "Uma entrada inválida modificou a lista."
    assert cadastrar_tarefa(tarefas, "  Ler livro  ") == "Tarefa cadastrada."
    assert tarefas[1]["titulo"] == "  Ler livro  ", "O título válido foi alterado."
    assert concluir_tarefa(tarefas, 0) == "Tarefa concluída."
    assert tarefas[0]["concluida"] is True
    assert tarefas[1]["concluida"] is False
    assert listar_tarefas(tarefas) == "0: [x] Comprar pão\n1: [ ]   Ler livro  "
    for indice in (-1, 2):
        antes = deepcopy(tarefas)
        assert concluir_tarefa(tarefas, indice) == "Tarefa não encontrada."
        assert tarefas == antes
    print("Verificações concluídas: cadastro, validação, listagem e conclusão.")



verificar()
```
